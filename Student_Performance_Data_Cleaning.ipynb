```python
# Import required libraries
import pandas as pd
import numpy as np
```

```python
# Upload the dataset
from google.colab import files
uploaded = files.upload()
```

```python
# Load the uploaded CSV file into a DataFrame
df = pd.read_csv(next(iter(uploaded)))

print("Dataset loaded successfully!")
```

```python
# Explore the dataset

print("First 5 rows:")
display(df.head())

print("\nDataset shape:")
print(df.shape)

print("\nColumn names:")
print(df.columns)

print("\nData types:")
print(df.dtypes)
```

```python
# Check for missing values in each column
df.isnull().sum()
```

```python
# Check the number of missing values before filling
print("Missing values before handling:")
print(df.isnull().sum())
```

```python
# Fill missing values using the median of each numeric column
df.fillna(df.median(numeric_only=True), inplace=True)

print("Missing values handled successfully!")
```

```python
# Verify that missing values have been handled
df.isnull().sum()
```

```python
# Check the data types of all columns
df.dtypes
```

```python
# Convert Assignments_Completed from float to integer
df["Assignments_Completed"] = df["Assignments_Completed"].astype(int)

# Check the updated data type
print(df["Assignments_Completed"].dtype)
```

```python
# Check for duplicate rows
print("Number of duplicate rows:")
print(df.duplicated().sum())
```

```python
# Remove duplicate rows
df = df.drop_duplicates()

print("Duplicate rows removed successfully!")
print("Current dataset shape:", df.shape)
```

```python
# Perform a statistical sanity check
df.describe()
```

```python
# Create Features (X) and Target (y)

X = df.drop(columns=["Result", "Final_Marks"])
y = df["Result"]

print("Features (X):")
print(X.columns)

print("\nTarget (y):")
print(y.name)
```

### Data Leakage

Data leakage occurs when information that should not be available at prediction time is given to the machine learning model.

In this dataset, Final_Marks should not be included in Features (X) because the Result is based on the student's final performance. Using Final_Marks as an input can give the model information that is too closely related to the target.

This can lead to artificially high model performance and unreliable predictions.

Therefore:
- X → input features available before the result is known
- y → Result, which the model needs to predict
- Final_Marks → excluded from X to avoid data leakage
